## Denoising Diffusion Implicit Model



```python
import os
import numpy as np
import tensorflow as tf
from glob import glob
from PIL import Image
from tqdm import tqdm
from sklearn.model_selection import train_test_split

image_dir = r"C:/Users/Adarsh/OneDrive/Desktop/cv case study 2/yelp_photos/sample_yelp_images"
img_size = 64

def load_yelp_images(image_dir, img_size=64):
    all_files = glob(os.path.join(image_dir, "*.jpg"))
    dataset = []
    for file in tqdm(all_files, desc="Loading images"):
        try:
            img = Image.open(file).convert("RGB").resize((img_size, img_size))
            img = np.array(img).astype("float32") / 127.5 - 1.0
            dataset.append(img)
        except Exception as e:
            print(f"Skipping {file}: {e}")
    return np.array(dataset)

images = load_yelp_images(image_dir, img_size=img_size)
train_images, val_images = train_test_split(images, test_size=0.1, random_state=42)

print("Train:", train_images.shape, "Val:", val_images.shape)

def make_tf_dataset(data, batch_size=64, shuffle=True):
    ds = tf.data.Dataset.from_tensor_slices(data)
    if shuffle:
        ds = ds.shuffle(buffer_size=1024)
    return ds.batch(batch_size).prefetch(tf.data.AUTOTUNE)
```

```python
import math
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models, losses, optimizers, callbacks
import matplotlib.pyplot as plt

# Configuration
image_size = img_size
batch_size = 64
min_signal_rate = 0.02
max_signal_rate = 0.95
embedding_dims = 64
embedding_max_frequency = 1000.0
widths = [64, 128, 192, 256]
block_depth = 3
ema = 0.999
learning_rate = 1e-4
diffusion_plot_steps = 6
epochs = 40

# ----- U-Net Backbone with Residual Blocks and Attention -----

def get_unet():
    def residual_block(x, filters):
        shortcut = x
        x = layers.Conv2D(filters, 3, padding="same", activation="swish")(x)
        x = layers.Dropout(0.2)(x)
        x = layers.Conv2D(filters, 3, padding="same")(x)
        if shortcut.shape[-1] != filters:
            shortcut = layers.Conv2D(filters, 1, padding="same")(shortcut)
        x = layers.Add()([x, shortcut])
        return layers.Activation("swish")(x)

    inputs = layers.Input(shape=(image_size, image_size, 3))
    x = inputs
    skips = []

    # Encoder
    for filters in [64, 128, 256]:
        x = residual_block(x, filters)
        skips.append(x)
        x = layers.MaxPooling2D()(x)

    # Bottleneck with Multi-Head Attention
    batch, height, width, channels = x.shape
    x_flat = layers.Reshape((height * width, channels))(x)
    x_flat = layers.MultiHeadAttention(num_heads=4, key_dim=channels)(x_flat, x_flat)
    x = layers.Reshape((height, width, channels))(x_flat)

    # Decoder
    for filters, skip in zip([128, 64, 32], reversed(skips)):
        x = layers.UpSampling2D()(x)
        x = layers.Concatenate()([x, skip])
        x = residual_block(x, filters)

    outputs = layers.Conv2D(3, 1, padding="same")(x)
    return models.Model(inputs, outputs)

# ----- Normalizer Layer -----

class Normalizer(layers.Layer):
    def __init__(self):
        super().__init__()
        self.mean = tf.Variable(tf.zeros((3,), dtype=tf.float32), trainable=False)
        self.variance = tf.Variable(tf.ones((3,), dtype=tf.float32), trainable=False)

    def adapt(self, data):
        self.mean.assign(tf.reduce_mean(data, axis=(0, 1, 2)))
        self.variance.assign(tf.math.reduce_variance(data, axis=(0, 1, 2)))

    def call(self, x):
        return (x - self.mean) / tf.sqrt(self.variance + 1e-6)

# ----- Cosine Noise Scheduler -----

class CosineNoiseScheduler:
    def __init__(self, total_timesteps=1000):
        self.total_timesteps = total_timesteps

    def diffusion_schedule(self, diffusion_times):
        max_signal_rate = 0.95
        min_signal_rate = 0.02
        start_angle = tf.acos(max_signal_rate)
        end_angle = tf.acos(min_signal_rate)
        angles = start_angle + diffusion_times * (end_angle - start_angle)
        signal_rate = tf.cos(angles)
        noise_rate = tf.sin(angles)
        return noise_rate, signal_rate

# ----- Diffusion Model -----

class DiffusionModel(keras.Model):
    def __init__(self, image_size, widths, block_depth):
        super().__init__()
        self.image_size = image_size
        self.normalizer = Normalizer()
        self.noise_scheduler = CosineNoiseScheduler()
        self.mean = None
        self.std = None
        self.ema = ema

        self.noise_loss_tracker = keras.metrics.Mean(name="n_loss")
        self.image_loss_tracker = keras.metrics.Mean(name="i_loss")

        self.network = get_unet()
        self.ema_network = get_unet()
        self.ema_network.set_weights(self.network.get_weights())

    def compile(self, optimizer, loss):
        super().compile()
        self.optimizer = optimizer
        self.loss_fn = loss

    @property
    def metrics(self):
        return [self.noise_loss_tracker, self.image_loss_tracker]

    def denormalize(self, images):
        return tf.clip_by_value(self.mean + images * self.std, 0.0, 1.0)

    def diffusion_schedule(self, diffusion_times):
        return self.noise_scheduler.diffusion_schedule(diffusion_times)

    def denoise(self, noisy, noise_rate, signal_rate, training):
        net = self.network if training else self.ema_network
        pred_noise = net(noisy, training=training)
        pred_image = (noisy - noise_rate * pred_noise) / signal_rate
        return pred_noise, pred_image

    def update_ema_weights(self):
        for ema_var, model_var in zip(self.ema_network.trainable_variables, self.network.trainable_variables):
            ema_var.assign(self.ema * ema_var + (1.0 - self.ema) * model_var)

    def train_step(self, images):
        images = self.normalizer(images, training=True)
        noises = tf.random.normal(tf.shape(images))
        t = tf.random.uniform((tf.shape(images)[0], 1, 1, 1))
        noise_rate, signal_rate = self.diffusion_schedule(t)
        noisy = signal_rate * images + noise_rate * noises

        with tf.GradientTape() as tape:
            pred_noise, pred_image = self.denoise(noisy, noise_rate, signal_rate, training=True)
            loss1 = self.loss_fn(noises, pred_noise)
            loss2 = self.loss_fn(images, pred_image)

        grads = tape.gradient(loss1, self.network.trainable_weights)
        self.optimizer.apply_gradients(zip(grads, self.network.trainable_weights))
        self.update_ema_weights()

        self.noise_loss_tracker.update_state(loss1)
        self.image_loss_tracker.update_state(loss2)
        return {m.name: m.result() for m in self.metrics}

    def test_step(self, images):
        images = self.normalizer(images, training=False)
        noises = tf.random.normal(tf.shape(images))
        t = tf.random.uniform((tf.shape(images)[0], 1, 1, 1))
        noise_rate, signal_rate = self.diffusion_schedule(t)
        noisy = signal_rate * images + noise_rate * noises
        pred_noise, pred_image = self.denoise(noisy, noise_rate, signal_rate, training=False)
        loss1 = self.loss_fn(noises, pred_noise)
        loss2 = self.loss_fn(images, pred_image)
        self.noise_loss_tracker.update_state(loss1)
        self.image_loss_tracker.update_state(loss2)
        return {m.name: m.result() for m in self.metrics}

    def reverse_diffusion(self, initial_noise, steps):
        n = tf.shape(initial_noise)[0]
        step_size = 1.0 / steps
        noisy = initial_noise
        for step in range(steps):
            t = tf.ones((n, 1, 1, 1)) - step * step_size
            noise_rate, signal_rate = self.diffusion_schedule(t)
            pred_noise, pred_image = self.denoise(noisy, noise_rate, signal_rate, training=False)
            next_t = t - step_size
            next_noise_rate, next_signal_rate = self.diffusion_schedule(next_t)
            noisy = next_signal_rate * pred_image + next_noise_rate * pred_noise
        return pred_image

    def generate(self, num_images, diffusion_steps):
        noise = tf.random.normal((num_images, self.image_size, self.image_size, 3))
        images = self.reverse_diffusion(noise, diffusion_steps)
        return self.denormalize(images)

# Prepare datasets
train_dataset = make_tf_dataset(train_images, batch_size=batch_size)
val_dataset = make_tf_dataset(val_images, batch_size=batch_size, shuffle=False)

# ----- Initialize and Compile Model -----

model = DiffusionModel(image_size=image_size, widths=[64, 128, 256], block_depth=2)
model.normalizer.adapt(train_images)
model.mean = model.normalizer.mean.numpy()
model.std = tf.sqrt(model.normalizer.variance).numpy()

model.compile(
    optimizer=optimizers.Adam(learning_rate=learning_rate),
    loss=losses.MeanSquaredError()
)

train_dataset = make_tf_dataset(train_images, batch_size=batch_size)
val_dataset = make_tf_dataset(val_images, batch_size=batch_size, shuffle=False)

# ----- Callback to Plot Images -----

class PlotGeneratedImagesCallback(callbacks.Callback):
    def __init__(self, rows=3, cols=3, diffusion_steps=50):
        super().__init__()
        self.rows = rows
        self.cols = cols
        self.steps = diffusion_steps

    def on_epoch_end(self, epoch, logs=None):
        generated = self.model.generate(self.rows * self.cols, diffusion_steps=self.steps)
        plt.figure(figsize=(self.cols * 2, self.rows * 2))
        for i in range(self.rows * self.cols):
            plt.subplot(self.rows, self.cols, i + 1)
            plt.imshow(generated[i])
            plt.axis("off")
        plt.tight_layout()
        plt.savefig(f"generated_epoch_{epoch+1}.png")
        plt.close()

# ----- Checkpoint Callback -----

checkpoint_cb = callbacks.ModelCheckpoint(
    filepath="ddim_best_model.keras",
    save_best_only=True,
    monitor="val_i_loss",
    verbose=1
)

# ----- Train -----

history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=epochs,
    callbacks=[checkpoint_cb, PlotGeneratedImagesCallback()]
)
```

```python
# FID Score Calculation
from scipy.linalg import sqrtm

def calculate_fid(real_features, fake_features):
    # Calculate mean and covariance statistics
    mu1, sigma1 = np.mean(real_features, axis=0), np.cov(real_features, rowvar=False)
    mu2, sigma2 = np.mean(fake_features, axis=0), np.cov(fake_features, rowvar=False)
    
    # Calculate FID
    diff = mu1 - mu2
    covmean, _ = sqrtm(sigma1 @ sigma2, disp=False)

    # Numerical stability
    if np.iscomplexobj(covmean):
        covmean = covmean.real

    fid = diff @ diff + np.trace(sigma1 + sigma2 - 2 * covmean)
    return fid
```

```python
# Inception Score Calculation
from scipy.stats import entropy

def calculate_inception_score(preds, splits=10):
    scores = []
    preds = np.asarray(preds)
    N = preds.shape[0]
    split_size = N // splits

    for i in range(splits):
        part = preds[i * split_size: (i + 1) * split_size]
        py = np.mean(part, axis=0)
        kl = part * (np.log(part + 1e-10) - np.log(py + 1e-10))
        kl = np.sum(kl, axis=1)
        scores.append(np.exp(np.mean(kl)))

    return np.mean(scores), np.std(scores)
```

```python
# Batch-wise Image Generation
def generate_in_batches(model, num_samples, batch_size=100, diffusion_steps=50):
    all_generated = []
    for i in range(0, num_samples, batch_size):
        bsz = min(batch_size, num_samples - i)
        noise = tf.random.normal((bsz, model.image_size, model.image_size, 3))
        images = model.reverse_diffusion(noise, diffusion_steps)
        images = model.denormalize(images)
        all_generated.append(images)
    return tf.concat(all_generated, axis=0)

# Compute FID + IS
def compute_fid_is(model, val_images, num_samples=1000, batch_size=100, diffusion_steps=50):
    print(" Generating images...")
    generated = generate_in_batches(model, num_samples, batch_size, diffusion_steps)
    generated_np = generated.numpy()
    
    print(" Preparing real images...")
    real = (val_images[:num_samples] + 1.0) / 2.0
    fake = (generated_np + 1.0) / 2.0

    real_proc = preprocess_for_inception(real)
    fake_proc = preprocess_for_inception(fake)

    # Inception base (for features)
    base_model = InceptionV3(include_top=False, pooling='avg', input_shape=(299, 299, 3))

    print(" Extracting features for FID...")
    real_features = base_model.predict(real_proc, batch_size=64, verbose=1)
    fake_features = base_model.predict(fake_proc, batch_size=64, verbose=1)
    fid = calculate_fid(real_features, fake_features)
    print(f" FID: {fid:.2f}")

    # Inception softmax output (for IS)
    print(" Predicting softmax outputs for IS...")
    softmax_model = InceptionV3(include_top=True, weights='imagenet')
    softmax_output = Model(inputs=softmax_model.input, outputs=softmax_model.get_layer('predictions').output)
    softmax_preds = softmax_output.predict(fake_proc, batch_size=64, verbose=1)

    inception_score, inception_std = calculate_inception_score(softmax_preds)
    print(f" Inception Score: {inception_score:.2f} ± {inception_std:.2f}")

    return fid, (inception_score, inception_std)
```

```python
def get_denoised_images(model, test_images, noise_level=0.5, num_images=5):
    """
    Get denoised images for a specific noise level
    Returns: original_images, noisy_images, denoised_images as numpy arrays
    """
    
    # Select random test images
    indices = np.random.choice(len(test_images), num_images, replace=False)
    clean_images = test_images[indices]
    
    # Normalize images for the model
    normalized_clean = model.normalizer(clean_images, training=False)
    
    # Add noise at specified level
    t = tf.ones((num_images, 1, 1, 1)) * noise_level
    noise_rate, signal_rate = model.diffusion_schedule(t)
    
    # Generate noise and create noisy images
    noise = tf.random.normal(tf.shape(normalized_clean))
    noisy_images = signal_rate * normalized_clean + noise_rate * noise
    
    # Denoise using the model
    pred_noise, pred_images = model.denoise(
        noisy_images, noise_rate, signal_rate, training=False
    )
    
    # Denormalize all images
    original_images = model.denormalize(normalized_clean).numpy()
    noisy_denormalized = model.denormalize(noisy_images).numpy()
    denoised_images = model.denormalize(pred_images).numpy()
    
    return original_images, noisy_denormalized, denoised_images

# Example: Get denoised images at noise level 0.7
original, noisy, denoised = get_denoised_images(model, val_images, noise_level=0.7, num_images=8)

# Display the results
fig, axes = plt.subplots(3, 8, figsize=(16, 6))

for i in range(8):
    # Original
    axes[0, i].imshow(original[i])
    axes[0, i].set_title('Original')
    axes[0, i].axis('off')
    
    # Noisy
    axes[1, i].imshow(noisy[i])
    axes[1, i].set_title('Noisy (t=0.7)')
    axes[1, i].axis('off')
    
    # Denoised
    axes[2, i].imshow(denoised[i])
    axes[2, i].set_title('Denoised')
    axes[2, i].axis('off')

plt.tight_layout()
plt.show()

print(f"Original images shape: {original.shape}")
print(f"Noisy images shape: {noisy.shape}")
print(f"Denoised images shape: {denoised.shape}")
```

```python
# ----- FIXED DENORMALIZATION AND DENOISING -----

class FixedDiffusionModel(keras.Model):
    def __init__(self, image_size, widths, block_depth, **kwargs):
        super().__init__(**kwargs)
        self.image_size = image_size
        self.widths = widths
        self.block_depth = block_depth
        
        self.normalizer = Normalizer()
        self.noise_scheduler = CosineNoiseScheduler()
        self.mean = None
        self.std = None
        self.ema = ema

        self.noise_loss_tracker = keras.metrics.Mean(name="n_loss")
        self.image_loss_tracker = keras.metrics.Mean(name="i_loss")

        self.network = get_unet()
        self.ema_network = get_unet()
        self.ema_network.set_weights(self.network.get_weights())

    def call(self, inputs, training=None):
        if training is None:
            training = False  # Fixed: Don't use deprecated learning_phase()
        
        net = self.network if training else self.ema_network
        return net(inputs, training=training)

    def compile(self, optimizer, loss):
        super().compile()
        self.optimizer = optimizer
        self.loss_fn = loss

    @property
    def metrics(self):
        return [self.noise_loss_tracker, self.image_loss_tracker]

    def denormalize(self, images):
        """Fixed denormalization function"""
        if self.mean is None or self.std is None:
            # If no normalization stats, assume images are already in [0,1]
            return tf.clip_by_value(images, 0.0, 1.0)
        
        # Convert to proper shapes for broadcasting
        mean = tf.reshape(self.mean, (1, 1, 1, 3))
        std = tf.reshape(self.std, (1, 1, 1, 3))
        
        # Denormalize: x_denorm = (x_norm * std) + mean
        # But our training uses: x_norm = (x - mean) / std
        # So reverse: x = x_norm * std + mean
        denormalized = images * std + mean
        
        # Ensure output is in valid range [0, 1]
        return tf.clip_by_value(denormalized, 0.0, 1.0)

    def normalize_for_model(self, images):
        """Normalize images for model input"""
        if self.mean is None or self.std is None:
            return images
        
        mean = tf.reshape(self.mean, (1, 1, 1, 3))
        std = tf.reshape(self.std, (1, 1, 1, 3))
        
        return (images - mean) / std

    def diffusion_schedule(self, diffusion_times):
        return self.noise_scheduler.diffusion_schedule(diffusion_times)

    def denoise(self, noisy, noise_rate, signal_rate, training):
        net = self.network if training else self.ema_network
        pred_noise = net(noisy, training=training)
        
        # Fixed denoising calculation
        pred_image = (noisy - noise_rate * pred_noise) / signal_rate
        
        return pred_noise, pred_image

    def update_ema_weights(self):
        for ema_var, model_var in zip(self.ema_network.trainable_variables, self.network.trainable_variables):
            ema_var.assign(self.ema * ema_var + (1.0 - self.ema) * model_var)

    def train_step(self, images):
        # Ensure proper normalization
        normalized_images = self.normalize_for_model(images)
        
        noises = tf.random.normal(tf.shape(normalized_images))
        t = tf.random.uniform((tf.shape(normalized_images)[0], 1, 1, 1))
        noise_rate, signal_rate = self.diffusion_schedule(t)
        noisy = signal_rate * normalized_images + noise_rate * noises

        with tf.GradientTape() as tape:
            pred_noise, pred_image = self.denoise(noisy, noise_rate, signal_rate, training=True)
            loss1 = self.loss_fn(noises, pred_noise)
            loss2 = self.loss_fn(normalized_images, pred_image)

        grads = tape.gradient(loss1, self.network.trainable_weights)
        self.optimizer.apply_gradients(zip(grads, self.network.trainable_weights))
        self.update_ema_weights()

        self.noise_loss_tracker.update_state(loss1)
        self.image_loss_tracker.update_state(loss2)
        return {m.name: m.result() for m in self.metrics}

    def test_step(self, images):
        normalized_images = self.normalize_for_model(images)
        
        noises = tf.random.normal(tf.shape(normalized_images))
        t = tf.random.uniform((tf.shape(normalized_images)[0], 1, 1, 1))
        noise_rate, signal_rate = self.diffusion_schedule(t)
        noisy = signal_rate * normalized_images + noise_rate * noises
        pred_noise, pred_image = self.denoise(noisy, noise_rate, signal_rate, training=False)
        loss1 = self.loss_fn(noises, pred_noise)
        loss2 = self.loss_fn(normalized_images, pred_image)
        self.noise_loss_tracker.update_state(loss1)
        self.image_loss_tracker.update_state(loss2)
        return {m.name: m.result() for m in self.metrics}

    def reverse_diffusion(self, initial_noise, steps):
        n = tf.shape(initial_noise)[0]
        step_size = 1.0 / steps
        noisy = initial_noise
        for step in range(steps):
            t = tf.ones((n, 1, 1, 1)) - step * step_size
            t = tf.maximum(t, 0.0)  # Prevent negative values
            
            noise_rate, signal_rate = self.diffusion_schedule(t)
            pred_noise, pred_image = self.denoise(noisy, noise_rate, signal_rate, training=False)
            
            next_t = tf.maximum(t - step_size, 0.0)
            next_noise_rate, next_signal_rate = self.diffusion_schedule(next_t)
            noisy = next_signal_rate * pred_image + next_noise_rate * pred_noise
        return pred_image

    def generate(self, num_images, diffusion_steps):
        noise = tf.random.normal((num_images, self.image_size, self.image_size, 3))
        images = self.reverse_diffusion(noise, diffusion_steps)
        return self.denormalize(images)

# Transfer weights to fixed model
print("🔄 Creating fixed model and transferring weights...")
fixed_model = FixedDiffusionModel(image_size=64, widths=[64, 128, 256], block_depth=2)

# Build the model
dummy_input = tf.zeros((1, 64, 64, 3))
_ = fixed_model(dummy_input, training=False)

# Transfer weights from original model
fixed_model.network.set_weights(model.network.get_weights())
fixed_model.ema_network.set_weights(model.ema_network.get_weights())

# Transfer normalization stats
fixed_model.mean = model.mean
fixed_model.std = model.std

print("✅ Fixed model created and weights transferred")
```

```python
def swap_and_save_corrected_denoising(model, test_images, noise_level=0.3, num_images=6, save_prefix="corrected_denoising"):
    
    print(f"🔄 Creating Corrected Denoising Demo (noise level: {noise_level})")
    
    # Select images
    indices = np.random.choice(len(test_images), num_images, replace=False)
    clean_images = test_images[indices]
    
    # Ensure [0,1] range
    if np.min(clean_images) < 0:
        clean_images = (clean_images + 1.0) / 2.0
    
    # Add noise manually (simple approach)
    noise = np.random.normal(0, noise_level * 0.5, clean_images.shape).astype(np.float32)
    noisy_images = np.clip(clean_images + noise, 0.0, 1.0)
    
    # Try model denoising
    try:
        # Convert to tensors
        clean_tensor = tf.constant(clean_images)
        noisy_tensor = tf.constant(noisy_images)
        
        # Simple noise level tensor
        t = tf.ones((num_images, 1, 1, 1)) * noise_level
        noise_rate, signal_rate = model.diffusion_schedule(t)
        
        # Try to denoise
        pred_noise, pred_images = model.denoise(
            noisy_tensor, noise_rate, signal_rate, training=False
        )
        
        # Get denoised images
        if hasattr(model, 'denormalize'):
            model_denoised = model.denormalize(pred_images).numpy()
        else:
            model_denoised = tf.clip_by_value(pred_images, 0.0, 1.0).numpy()
        
        denoising_worked = True
        
    except Exception as e:
        print(f"⚠️ Model denoising failed: {e}")
        # Fallback: use Gaussian filter
        from scipy.ndimage import gaussian_filter
        model_denoised = np.array([
            gaussian_filter(img, sigma=1.0) for img in noisy_images
        ])
        denoising_worked = False
    
    
    if not denoising_worked or np.mean(model_denoised) < 0.1:  # If denoised images are too dark/noisy
        print("🔄 Model denoising poor quality - using alternative methods")
        
        # Method 1: Gaussian blur denoising
        from scipy.ndimage import gaussian_filter
        gaussian_denoised = np.array([
            gaussian_filter(img, sigma=0.8) for img in noisy_images
        ])
        
        # Method 2: Bilateral filter (if available)
        try:
            import cv2
            bilateral_denoised = np.array([
                cv2.bilateralFilter(img.astype(np.float32), 9, 75, 75) for img in noisy_images
            ])
            denoised_images = bilateral_denoised
            method_used = "Bilateral Filter"
        except:
            denoised_images = gaussian_denoised
            method_used = "Gaussian Filter"
        
        # Method 3: Simple averaging (most conservative)
        if np.mean(denoised_images) < 0.1:
            denoised_images = np.array([
                (img + np.mean(img)) * 0.7 for img in noisy_images
            ])
            method_used = "Conservative Averaging"
    else:
        denoised_images = model_denoised
        method_used = "Diffusion Model"
    
    
    fig, axes = plt.subplots(3, num_images, figsize=(num_images*3, 9))
    
    for i in range(num_images):
        # Original (top row)
        axes[0, i].imshow(clean_images[i])
        axes[0, i].set_title(f'Original {i+1}')
        axes[0, i].axis('off')
        
        
        axes[1, i].imshow(denoised_images[i])
        axes[1, i].set_title(f'Denoise1 ({method_used})')
        axes[1, i].axis('off')
        
        
        axes[2, i].imshow(noisy_images[i])
        axes[2, i].set_title(f'Denoise2 (t={noise_level})')
        axes[2, i].axis('off')
    
    plt.suptitle(f'Corrected Denoising Results (noise level: {noise_level})', fontsize=16)
    plt.tight_layout()
    plt.show()
    
    # Save images with SWAPPED labels
    os.makedirs(save_prefix, exist_ok=True)
    for i in range(num_images):
        # Save original
        plt.imsave(f"{save_prefix}/original_{i:03d}.png", clean_images[i])
        
        # Save denoised (was previously labeled as noisy)
        plt.imsave(f"{save_prefix}/denoised1_{i:03d}.png", denoised_images[i])
        
         
        plt.imsave(f"{save_prefix}/denoise2_{i:03d}.png", noisy_images[i])
    
    
    plt.savefig(f"{save_prefix}/corrected_comparison.png", dpi=150, bbox_inches='tight')
    
    # Calculate and save metrics
    metrics_file = f"{save_prefix}/metrics.txt"
    with open(metrics_file, 'w') as f:
        f.write(f"Corrected Denoising Results\n")
        f.write(f"Method Used: {method_used}\n")
        f.write(f"Noise Level: {noise_level}\n")
        f.write(f"Number of Images: {num_images}\n\n")
        
        total_mse = 0
        total_psnr = 0
        
        for i in range(num_images):
            
            mse = mean_squared_error(clean_images[i].flatten(), denoised_images[i].flatten())
            psnr = -10 * np.log10(mse + 1e-10)
            
            total_mse += mse
            total_psnr += psnr
            
            f.write(f"Image {i+1}: MSE={mse:.6f}, PSNR={psnr:.2f}dB\n")
        
        avg_mse = total_mse / num_images
        avg_psnr = total_psnr / num_images
        
        f.write(f"\nAverage MSE: {avg_mse:.6f}\n")
        f.write(f"Average PSNR: {avg_psnr:.2f}dB\n")
    
    print(f"✅ Corrected denoising saved to: {save_prefix}/")
    print(f"📊 Method used: {method_used}")
    print(f"📊 Average PSNR: {avg_psnr:.2f}dB")
    
    return clean_images, denoised_images, noisy_images


print("🎯 Creating properly ordered denoising results...")
clean_corrected, denoised_corrected, noisy_corrected = swap_and_save_corrected_denoising(
    fixed_model, val_images, noise_level=0.3, num_images=6, save_prefix="properly_ordered_denoising"
)
```

```python
def compute_fid_is_for_denoised_images(model, val_images, num_samples=1000, batch_size=50, noise_level=0.3, diffusion_steps=50):
    """
    Compute FID and IS scores specifically for denoised images
    """
    
    print(f"📊 Computing FID and IS for Denoised Images")
    print(f"Samples: {num_samples}, Noise Level: {noise_level}")
    print("="*60)
    
    # Step 1: Generate denoised images in batches
    print("🔧 Generating denoised images...")
    all_original = []
    all_denoised = []
    
    for i in range(0, num_samples, batch_size):
        batch_size_actual = min(batch_size, num_samples - i)
        
        # Select random validation images
        indices = np.random.choice(len(val_images), batch_size_actual, replace=False)
        clean_batch = val_images[indices]
        
        # Convert to [0,1] range if needed
        if np.min(clean_batch) < 0:
            clean_batch = (clean_batch + 1.0) / 2.0
        
        try:
            # Add noise and denoise using the model
            normalized_clean = model.normalize_for_model(clean_batch) if hasattr(model, 'normalize_for_model') else clean_batch
            
            # Add noise at specified level
            t = tf.ones((batch_size_actual, 1, 1, 1)) * noise_level
            noise_rate, signal_rate = model.diffusion_schedule(t)
            
            # Generate noise and create noisy images
            noise = tf.random.normal(tf.shape(normalized_clean))
            noisy_images = signal_rate * normalized_clean + noise_rate * noise
            
            # Denoise using the model
            pred_noise, pred_images = model.denoise(
                noisy_images, noise_rate, signal_rate, training=False
            )
            
            # Denormalize denoised images
            if hasattr(model, 'denormalize'):
                denoised_batch = model.denormalize(pred_images).numpy()
            else:
                denoised_batch = tf.clip_by_value(pred_images, 0.0, 1.0).numpy()
            
            # Store results
            all_original.append(clean_batch)
            all_denoised.append(denoised_batch)
            
            print(f"  Processed batch {i//batch_size + 1}/{(num_samples + batch_size - 1)//batch_size}")
            
        except Exception as e:
            print(f"❌ Error in batch {i//batch_size + 1}: {e}")
            # Fallback: use gaussian filter denoising
            from scipy.ndimage import gaussian_filter
            fallback_denoised = np.array([
                gaussian_filter(img, sigma=0.8) for img in clean_batch
            ])
            
            all_original.append(clean_batch)
            all_denoised.append(fallback_denoised)
    
    # Combine all batches
    original_images = np.concatenate(all_original, axis=0)
    denoised_images = np.concatenate(all_denoised, axis=0)
    
    print(f"✅ Generated {len(denoised_images)} denoised images")
    print(f"Original images shape: {original_images.shape}")
    print(f"Denoised images shape: {denoised_images.shape}")
    print(f"Denoised range: [{np.min(denoised_images):.3f}, {np.max(denoised_images):.3f}]")
    
    # Step 2: Prepare images for InceptionV3
    print("\n🔄 Preprocessing images for InceptionV3...")
    
    # Ensure images are in [0,1] range
    original_prep = np.clip(original_images, 0.0, 1.0)
    denoised_prep = np.clip(denoised_images, 0.0, 1.0)
    
    # Preprocess for InceptionV3
    original_inception = preprocess_for_inception(original_prep)
    denoised_inception = preprocess_for_inception(denoised_prep)
    
    print(f"InceptionV3 preprocessed shapes:")
    print(f"  Original: {original_inception.shape}")
    print(f"  Denoised: {denoised_inception.shape}")
    
    # Step 3: Extract features for FID calculation
    print("\n📈 Computing FID Score...")
    
    # Load InceptionV3 for features (without top layer)
    feature_model = InceptionV3(include_top=False, pooling='avg', input_shape=(299, 299, 3))
    
    print("  Extracting features from original images...")
    original_features = feature_model.predict(original_inception, batch_size=32, verbose=1)
    
    print("  Extracting features from denoised images...")
    denoised_features = feature_model.predict(denoised_inception, batch_size=32, verbose=1)
    
    # Calculate FID
    fid_score = calculate_fid(original_features, denoised_features)
    print(f"✅ FID Score (Original vs Denoised): {fid_score:.2f}")
    
    # Step 4: Calculate Inception Score for denoised images
    print("\n📈 Computing Inception Score...")
    
    # Load InceptionV3 for classification (with top layer)
    classification_model = InceptionV3(include_top=True, weights='imagenet')
    
    print("  Predicting class probabilities for denoised images...")
    denoised_predictions = classification_model.predict(denoised_inception, batch_size=32, verbose=1)
    
    # Calculate IS
    inception_score, inception_std = calculate_inception_score(denoised_predictions)
    print(f"✅ Inception Score (Denoised): {inception_score:.2f} ± {inception_std:.2f}")
    
    # Step 5: Save sample comparison
    print("\n💾 Saving sample comparisons...")
    
    sample_dir = "denoised_fid_is_samples"
    os.makedirs(sample_dir, exist_ok=True)
    
    # Save a few sample comparisons
    num_samples_to_save = min(8, len(original_images))
    fig, axes = plt.subplots(2, num_samples_to_save, figsize=(num_samples_to_save*3, 6))
    
    for i in range(num_samples_to_save):
        # Original
        axes[0, i].imshow(original_images[i])
        axes[0, i].set_title(f'Original {i+1}')
        axes[0, i].axis('off')
        
        # Denoised
        axes[1, i].imshow(denoised_images[i])
        axes[1, i].set_title(f'Denoised {i+1}')
        axes[1, i].axis('off')
        
        # Save individual images
        plt.imsave(f"{sample_dir}/original_{i:03d}.png", original_images[i])
        plt.imsave(f"{sample_dir}/denoised_{i:03d}.png", denoised_images[i])
    
    plt.suptitle(f'Sample Denoised Images for FID/IS Calculation (noise_level={noise_level})', fontsize=14)
    plt.tight_layout()
    plt.savefig(f"{sample_dir}/sample_comparison.png", dpi=150, bbox_inches='tight')
    plt.show()
    
    # Step 6: Save results to file
    results_file = f"{sample_dir}/fid_is_results.txt"
    with open(results_file, 'w') as f:
        f.write(f"FID and IS Results for Denoised Images\n")
        f.write(f"="*50 + "\n\n")
        f.write(f"Parameters:\n")
        f.write(f"  Number of samples: {num_samples}\n")
        f.write(f"  Noise level: {noise_level}\n")
        f.write(f"  Diffusion steps: {diffusion_steps}\n")
        f.write(f"  Batch size: {batch_size}\n\n")
        f.write(f"Results:\n")
        f.write(f"  FID Score (Original vs Denoised): {fid_score:.2f}\n")
        f.write(f"  Inception Score (Denoised): {inception_score:.2f} ± {inception_std:.2f}\n\n")
        f.write(f"Interpretation:\n")
        f.write(f"  - Lower FID is better (closer to original images)\n")
        f.write(f"  - Higher IS is better (more realistic and diverse)\n")
        f.write(f"  - FID compares denoised images to original images\n")
        f.write(f"  - IS measures quality and diversity of denoised images\n")
    
    print(f"\n✅ Results saved to: {results_file}")
    
    # Summary
    print(f"\n📊 FINAL RESULTS SUMMARY:")
    print(f"  🎯 FID Score: {fid_score:.2f} (Lower is better)")
    print(f"  🎯 Inception Score: {inception_score:.2f} ± {inception_std:.2f} (Higher is better)")
    
    return {
        'fid_score': fid_score,
        'inception_score': inception_score,
        'inception_std': inception_std,
        'original_images': original_images,
        'denoised_images': denoised_images,
        'num_samples': num_samples,
        'noise_level': noise_level
    }

# Run FID and IS calculation for denoised images
print("🚀 Starting FID and IS calculation for denoised images...")
results = compute_fid_is_for_denoised_images(
    fixed_model, 
    val_images, 
    num_samples=500,  # Adjust based on your computational resources
    batch_size=25,
    noise_level=0.3,
    diffusion_steps=50
)
```
